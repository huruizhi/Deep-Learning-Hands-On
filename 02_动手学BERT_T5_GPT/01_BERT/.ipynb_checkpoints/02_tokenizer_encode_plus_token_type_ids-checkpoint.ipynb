## tokenizer 补充
- input_ids, attention_mask
  - mask：bert 另外一个预训练任务，mlm；
- encode_plus, token_type_ids
- token_type_ids: 0：表示第一句，1：表示第二句；可以通过 tokenizer()（tokenizer.__call__：都是0的）；也可以通过 encode_plus 生成/返回（前一句为0，后一句为1）；
  - 句子对，一般使用在 nsp（next sentence predict，bert 预训练任务）

```python
token_news = newsgroups_train.data[:3]
len(token_news[0])
```

```python
tokenizer(token_news[0],  truncation=True, padding=True, return_tensors="pt")
```

```python
# 句子对
tokenizer.encode_plus(text=token_news[0],text_pair=token_news[1])
```
